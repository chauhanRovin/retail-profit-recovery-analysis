# Retail Profit Recovery Analysis

## Notebook 2 – Data Cleaning

### Project Overview

Before analyzing business performance, it is important to ensure that the dataset is clean, consistent, and reliable.

Poor data quality can lead to incorrect conclusions and poor business decisions. In this notebook, the dataset will be examined for common data quality issues and prepared for analysis.

## Business Objective

The objectives of this notebook are to:

- Improve data quality.
- Correct data type issues.
- Remove duplicate records if present.
- Handle missing values.
- Prepare the dataset for business analysis.

## Step 1: Load the Dataset

The cleaned workflow begins by loading the original dataset used during the data understanding phase.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("/content/superstore.csv")
```

## Step 2: Create a Working Copy

A copy of the dataset is created to preserve the original data.

This is considered a good practice because it prevents accidental modification of the raw dataset.

```python
df_clean = df.copy()
```

## Step 3: Check Missing Values

Missing values are reviewed to determine whether any columns require cleaning before analysis.

```python
df_clean.isnull().sum()
```

### Observation

No missing values were found.

The dataset is complete and does not require imputation.

## Step 4: Identify Duplicate Records

Duplicate records can distort business metrics such as total sales, profit, and customer counts.

The dataset is checked to determine whether duplicate transactions exist.

```python
df_clean.duplicated().sum()
```

### Observation

No duplicate records were identified.

## Step 5: Convert Date Columns

Date columns should be stored using the datetime format to enable time-series analysis such as monthly sales trends and seasonal performance.

```python
df_clean["Order.Date"] = pd.to_datetime(df_clean["Order.Date"])
df_clean["Ship.Date"] = pd.to_datetime(df_clean["Ship.Date"])
```

## Step 6: Verify Data Types

After converting the date columns, the data types are reviewed to confirm that all variables are stored correctly.

```python
df_clean.info()
```

## Step 7: Validate Numerical Columns

Key business metrics such as Sales, Profit, Quantity, Discount, and Shipping Cost are reviewed to ensure they contain valid numerical values.

```python
df_clean[["Sales","Profit","Quantity","Discount","Shipping.Cost"]].describe()
```

## Step 8: Save the Clean Dataset

The cleaned dataset is saved separately to preserve the original raw data.

All future analysis will use this cleaned version.

```python
import os

clean_folder = "/content/drive/MyDrive/Retail-Profit-Recovery-Analysis/data/cleaned"

os.makedirs(clean_folder, exist_ok=True)

df_clean.to_csv(f"{clean_folder}/superstore_clean.csv", index=False)
```

- Duplicate records checked
- Date columns converted to datetime format

- Clean dataset saved for future analysis
