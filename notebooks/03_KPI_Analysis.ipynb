## Notebook 3 – Executive KPI Analysis

### Project Overview

After cleaning the dataset, the next step is to evaluate the overall business performance using Key Performance Indicators (KPIs).

KPIs provide a high-level summary of business performance and help management understand the current state of the business before conducting detailed investigations.

This notebook focuses on calculating and interpreting the most important business metrics.

## Step 1 – Load the Clean Dataset

The cleaned dataset prepared in the previous notebook is loaded for KPI analysis.

Using the cleaned dataset ensures that the calculated business metrics are accurate and reliable.

```python
import pandas as pd
df = pd.read_csv("/content/superstore_clean.csv")
```

## KPI 1 – Total Sales

Total Sales represents the total revenue generated from all customer orders.

This KPI provides an overall view of the company's revenue performance.

```python
total_sales = df["Sales"].sum()
print(f"Total Sales : ${total_sales:,.2f}")
```

### Business Observation

This KPI represents the total revenue generated during the available period.

It serves as the primary indicator of business growth and market demand.

## KPI 2 – Total Profit

Total Profit measures the earnings remaining after deducting all product-related costs.

This KPI helps evaluate the company's profitability.

```python
total_profit = df["Profit"].sum()

print(f"Total Profit : ${total_profit:,.2f}")
```

```python
### Business Observation

Although revenue may be high, profitability determines whether the business is creating value.

This KPI will later be compared with Sales to understand overall business efficiency.
```

## KPI 3 – Profit Margin

Profit Margin indicates how much profit is earned for every dollar of sales.

It is one of the most important indicators of business health.

```python
profit_margin = (total_profit/total_sales)*100

print(f"Profit Margin : ${profit_margin:,.2f}")
```

### Business Observation

A higher profit margin indicates better operational efficiency and pricing strategy.

A low profit margin may suggest excessive discounts, high operational costs, or inefficient product pricing.

```python
#Total Orders
total_orders = df["Order.ID"].nunique()
print(f"Total Orders : ${total_orders:,}")
```

```python
#Total Customers
total_customers = df["Customer.ID"].nunique()

print("Total Customers:", total_customers)
```

```python
#Total Quantity Sold
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
#Average Order Value
average_order_value = total_sales/total_orders

print(f"Average Order Value : ${average_order_value:.2f}")
```

```python
kpi = pd.DataFrame({
    "KPI":[
        "Total Sales",
        "Total Profit",
        "Profit Margin",
        "Total Orders",
        "Total Customers",
        "Average Order Value",
        "Total Quantity"
    ],

    "Value":[
        total_sales,
        total_profit,
        profit_margin,
        total_orders,
        total_customers,
        average_order_value,
        total_quantity
    ]
})

kpi
```
